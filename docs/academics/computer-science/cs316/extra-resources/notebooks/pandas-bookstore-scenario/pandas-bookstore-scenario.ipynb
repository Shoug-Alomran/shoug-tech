os.makedirs("bookstore_data", exist_ok=True)

**`exist_ok`** is a boolean parameter for `os.makedirs()` that controls what happens if the target directory **already exists**:

- **`exist_ok=False`** (the default) — raises a `FileExistsError` if the directory already exists.

It's useful for exactly the situation in the notebook: if you run the cell once, it creates `bookstore_data/`. If you run the notebook again later (or re-run that cell), the folder is already there — without `exist_ok=True`, Python would crash with an error instead of just continuing.

## Part 1 — Pandas `Series`

Your manager first sends you a quick price list for five bestselling titles.
A `Series` is a **one-dimensional labeled array** — perfect for a single column of data.

### 1.1 Creating a Series — from a list (default integer index)

````python
# TODO: create a Series from this list of prices
prices = pd.Series([45.0, 60.0, 35.5, 80.0, 25.0])
print(prices)
print("\nType:", type(prices)) # in series each element in a list should have the same data type
````

### 1.2 Creating a Series — from a list, with a custom index

````python
titles = ["Atomic Habits", "Dune", "The Alchemist", "Sapiens", "1984"]

# TODO: create a Series of the same prices, but use `titles` as the index
book_prices = pd.Series([45.0, 60.0, 35.5, 80.0, 25.0], index=titles, name= 'prices') # Execute the same line without 'name='price', assign titles as the index to give it a meaningful label, if you didnt specify a label to the column it will automatically set it to 0
book_prices
````

### 1.3 Creating a Series — from a dictionary

Keys automatically become the index.

````python
stock_dict = {
    "Atomic Habits": 12,
    "Dune": 5,
    "The Alchemist": 20,
    "Sapiens": 8,
    "1984": 15,
}
# TODO: create a Series directly from the dictionary above
stock = pd.Series(stock_dict)
stock

# keys represent attributes in a dictionary that are descriptive of their value
# Example:
# Student_Dict = {"Student_ID": 223, "Name": "Layan"}
````

### 1.4 Accessing data in a Series

- **By label** — `series["label"]`
- **By position** — `series.iloc[position]`

````python
# TODO: access the price of "Dune" by its index LABEL
print("Price of Dune:", book_prices["Dune"])
# need to execute previous cells in order to define the required series

# TODO: access the FIRST book's price by POSITION (not label)
print("First book's price:", book_prices.iloc[0])
````

### 1.5 Slicing a Series

````python
# TODO: slice the first three books BY POSITION (end-exclusive, like a normal Python list)
# print("First three books:\n", book_prices.iloc[0:3]) # 0, 1, 2, where 3 is exclusive

# TODO: slice BY LABEL from "Dune" to "Sapiens" (end-INCLUSIVE for labels)
print("\nFrom 'Dune' to 'Sapiens':\n", book_prices[["Dune", "The Alchemist", "Sapiens"]])

# TODO: slice with a step to get every 2nd book (by position)
# print("\nEvery 2nd book:\n", book_prices.iloc[::2])
````

Quick Note


`book_prices.iloc[::2]` is **Pandas syntax for selecting every second row** from the Series `book_prices`.

Let's break it down:

### 1. `.iloc`

`.iloc` means **integer-location based indexing**. It selects rows/columns based on their numerical position, rather than their labels.

### 2. `[::2]`

This follows Python's slicing syntax:

```python
[start : stop : step]
```

Here:

```python
[::2]
```

means:

* **start:** omitted → start from the beginning
* **stop:** omitted → go until the end
* **step:** `2` → take every 2nd element

So:

```python
book_prices.iloc[::2]
```

means:

> **Select rows at positions 0, 2, 4 **



### Other useful examples

```python
book_prices.iloc[::1]   # every row
book_prices.iloc[::2]   # every 2nd row
book_prices.iloc[::3]   # every 3rd row
book_prices.iloc[1::2]  # every 2nd row, starting from position 1
book_prices.iloc[::-1]  # all rows, but in reverse order
```

A useful way to remember it is:

**`iloc[start : stop : step]` → position-based slicing.**


## Part 2 — Pandas `DataFrame`

A single price list isn't enough — you need prices **and** stock **and** genre together.
A `DataFrame` combines multiple `Series` (columns) that share the same index, into a
**two-dimensional labeled table**.

### 2.1 Two Series, combined into a DataFrame

````python

genres = pd.Series(
    ["Self-help", "Sci-Fi", "Fiction", "Non-fiction", "Dystopian"],
    index=titles,
)

# Combine Series -> DataFrame
# TODO: combine book_prices, stock, and genres into ONE DataFrame,
# with columns named "price", "stock", and "genre"
books_df = pd.DataFrame({
    "price": book_prices,
    "stock": stock,
    "genre": genres,
})
# try to remove the keys in the dataframes

# keys specify the column name
books_df
````

### 2.2 Creating a DataFrame directly — from a dictionary

````python
sales_df = pd.DataFrame({
    "title": ["Atomic Habits", "Dune", "Atomic Habits", "1984", "Sapiens", "Dune"],
    "branch": ["Olaya", "Olaya", "Tahlia", "Tahlia", "Olaya", "Tahlia"],
    "units_sold": [3, 1, 2, 4, 2, 1],
    "unit_price": [45.0, 60.0, 45.0, 25.0, 80.0, 60.0],
})
# TODO: add a new column "revenue" = units_sold * unit_price
sales_df["revenue"] = sales_df["units_sold"] * sales_df["unit_price"] # new col deduced from units sold and unit price
sales_df
````

### 2.3 Accessing data in a DataFrame

- **Column(s):** `df["col"]` or `df[["col1", "col2"]]`
- **Rows by position:** `df.iloc[...]`
- **Rows by label:** `df.loc[...]`

````python
# TODO: select the single column "title" (returns a Series) and print its first 5 rows
print(sales_df["title"].head()) # head() retrieves the first 5 rows

# TODO: select BOTH "title" and "revenue" columns (returns a DataFrame)
sales_df[["title", "revenue"]].head()
````

````python
# Rows by position
# TODO: get the first row by POSITION
print(sales_df.iloc[0])
print()

# TODO: get rows 1 through 2 by POSITION (end-exclusive)
print(sales_df.iloc[1:3])
````

## Part 3 — Indexing and Selection

Your manager wants two specific reports. This is where `.loc[]` (label-based) and
`.iloc[]` (integer/position-based) come in.

### 3.1 Label-based indexing with `.loc[]`

````python
# TODO: select ALL rows, but only the "title", "branch", and "revenue" columns, using .loc[]
select = sales_df.loc[:, ["title","branch","revenue"]] # loc() takes two parameters: 1. specifies rows 2. specifies column
# prints all rows
print(select)
# select structure contains all rows but here we want to display 2 rows
select.head(2)
````

````python
# TODO: conditional selection — only rows where branch == "Olaya"
sales_df.loc[sales_df["branch"] =="Olaya"]
````

````python
# TODO: filter by TWO conditions: branch is "Olaya" AND revenue > 100
# Remember: combine conditions with & (AND) or | (OR), each condition in parentheses
sales_df.loc[(sales_df["branch"]=="Olaya") & (sales_df["revenue"]>100)] # NOTE: and operator is denoted by &
````

### 3.2 Integer-based indexing with `.iloc[]`

````python
# TODO: select rows 0-2 and columns 0-1 (position based, end-exclusive) using .iloc[]
sales_df.iloc[0:3,0:2]
````

## Part 4 — Reading Data into Pandas

In reality, your manager's data doesn't arrive as neat Python dictionaries — it arrives
as **CSV**, **Excel**, **SQL**, and **JSON** files. Let's simulate that: first we export
`sales_df` in each format (Part 5 goes into this in more depth), then read each one back
in, exactly as you would with real files from a colleague.

````python
# Quick exports so we have real files to read from (details in Part 5)
sales_df.to_csv("bookstore_data/sales.csv", index=False) # 'to_csv' writes the dataframe into a .CSV file - file sales stored in folder -> for later use - method to_csv passes the directory path
sales_df.to_excel("bookstore_data/sales.xlsx", sheet_name="Sales", index=True) # specifies sheet names if the spreadsheet contains multiple sheets
sales_df.to_json("bookstore_data/sales.json", orient="records") # orient parameter takes one of two values: records (row by row) or columns (column by column)
sales_df.to_json("bookstore_data/sales_columns.json", orient="columns")
import sqlite3 # built-in library used to work with databases
conn = sqlite3.connect("bookstore_data/bookstore.db") # establishes a connection to the database
sales_df.to_sql("sales", conn, if_exists="replace", index=False) # 1. name of the table 2. connect 3. replace the existing db
conn.close()

print("Files created:", os.listdir("bookstore_data"))
````

## Part 5 — Exporting Data from Pandas

Now imagine you've cleaned and summarized the data — your manager wants **outputs**
in different formats for different audiences.

### 5.1 Exporting to CSV (with common parameters)

````python
summary = (
    sales_df.groupby("branch", as_index=False)["revenue"].sum().rename(columns={"revenue": "total_revenue"})
)

# TODO: basic export to "bookstore_data/branch_summary.csv" WITHOUT the index column
summary.to_csv("bookstore_data/branch_summary.csv", index=False)

# TODO: advanced export — same data, but use ";" as the separator and "utf-8" encoding,
# saved to "bookstore_data/branch_summary_semicolon.csv" -> encoding - we can remove and by default it will run
summary.to_csv("bookstore_data/branch_summary_semicolon.csv", sep=";", encoding="utf-8", index=False)

summary
````

### 5.2 Exporting to Excel

````python
# TODO: export `summary` to "bookstore_data/branch_summary.xlsx", sheet name "Summary", no index column
summary.to_excel("bookstore_data/branch_summary.xlsx", sheet_name="Summary", index=False)
print("Saved branch_summary.xlsx")
````

### 5.3 Exporting to SQL

````python
conn = sqlite3.connect("bookstore_data/bookstore.db")
# TODO: write `summary` to a table called "branch_summary", replacing it if it already exists
summary.to_sql("branch_summary", conn, if_exists="replace", index=False)


# Confirm it round-trips
check = pd.read_sql("SELECT * FROM branch_summary", conn)
conn.close()
check
````

### 5.4 Exporting to JSON — comparing `orient` options

````python
# TODO: export `summary` to JSON twice — once with orient="records", once with orient="columns"
summary.to_json("bookstore_data/summary_records.json", orient="records")
summary.to_json("bookstore_data/summary_columns.json", orient="columns")

with open("bookstore_data/summary_records.json") as f:
    print("orient='records':\n", f.read())

with open("bookstore_data/summary_columns.json") as f:
    print("\norient='columns':\n", f.read())
````
