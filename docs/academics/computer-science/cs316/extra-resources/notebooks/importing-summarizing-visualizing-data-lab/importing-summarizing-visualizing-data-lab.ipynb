## 17. Boxplot for a Quantitative Variable
Summarize the distribution of the `age` feature using a horizontal boxplot.

```python
#TODO - visualized ONE quantitative variable
plt.boxplot(nutri['age'], widths=width, vert=False)
# Selects age column, width controls thickness of the box (visual formatting), vert controls orientation - vertical/horizontally boxplot
plt.xlabel('age') # if orientation is horizontal it should run on the x-axis and vice versa
plt.show()

# Boxplot is mainly used to represent examine the distribution/spread of one quantitative var across the entire dataset
# Specifically, the box represents the median of the data
```

## 18. Histogram
Visualize the age distribution of the elderly participants as a proportion-weighted histogram.

```python
weights = np.ones_like(nutri.age) / nutri.age.count()
# return existing ages as 1s and divide it by the total observation
# 1/200 = 0.005 - used to show proportions of each age group rather than its raw count
# if 40 ppl fall into an age interval, the bar height =40
# with proportion weights = 0.20, thus 20% of the bar represents this observation

#TODO
plt.hist(nutri['age'], bins=9, weights=weights, facecolor='cyan',
         edgecolor='black', linewidth=0.4)
# nutri['age']= quantitative data being plotted
# bins=9 -> divide all age ranges into 9 intervals
# weights -> demonstrates proportions rather than raw frequency of ages
# facecolor -> customizes the color of the bar
# edgecolor -> customizes the border color
# linewidth -> thickness ratio of the border
plt.xlabel('age')
plt.ylabel('Proportion of Total')
plt.show()

# Differentiate - unlike bar graphs where each bar represents distinct categories, histograms' represent numerical intervals that visualize approximation of most common age groups in this context
```

## 19. Empirical Cumulative Distribution Function (ECDF)
Plot the empirical CDF of the `age` variable, showing the proportion of observations less than or equal to each value.

```python
x = np.sort(nutri.age) # sorts observations
y = np.linspace(0, 1, len(nutri.age)) # generates as many evenly spaced no as there are ages
plt.xlabel('age')
plt.ylabel('Fn(x)') # proportions of observations <= x - representing age
plt.step(x, y) # plot ages against cumulative proportions / proportions accumulating to it as a staircase structure
plt.xlim(x.min(), x.max()) # lower and upper bound of ages
plt.show()
```

## 20. Bivariate Barplot (Two Categorical Variables)
Compare barplots of `situation` split by `gender` using grouped/subplot bar charts.

```python
#fig,ax = plt.subplots() # (2,2, figsize=12,4)
ct = pd.crosstab(nutri.situation, nutri.gender) # used to combine/intersect two categorical variables
ct.plot(kind='bar', edgecolor='black') # plots cross-tab as grouped bar i.e. each bar is composed of two groups based on the crosstab operation
plt.ylabel('Counts')
plt.xlabel('Categories')
plt.show()

# ask about fig, ax
# fig- represents the entire figure that holds the data plots
# ax- represents the individual plots drawn within the figure
# difference between usinf fig/ax to not using it is automatic vs. explicit control
# without it, panda by default creates the axes and figures
# with it, you can explicitly instruct it to tell pandas precisely where to draw the plot
# useful for multiple plots within a figure
# not particularly useful but rather redundant
```

## 21. Same Bivariate Plot with Seaborn
Reproduce the grouped barplot directly from the DataFrame using seaborn's `countplot`, without manually computing the counts first.

```python
import seaborn as sns

sns.countplot(x='situation', hue='gender', data=nutri,
              hue_order=['Male', 'Female'],
              palette=['SkyBlue', 'Pink'],
              saturation=2, edgecolor='black')
plt.legend(loc='upper center') # inserts the legend in a specific location e.g. upper right/left/center
plt.xlabel('Situation Categories')
plt.ylabel('Counts')
plt.show()

# x parameter: defines the main categories along x-axis
# hue: variable that will split each situation category e.g. gender
# data: injects the dataframe that seaborn can refer to extract the required variables
# hue_order: controls the order of each gender group
# palette: controls the colors of the hue_order
# saturation: controls color intensity
# legend - defines which bar color corresponds to which gender group
```

```python
# A university wants to visualize how many students belong to CS, IS, SE, and Cybersecurity.
# Variable: Qualitative
# One Var
# Compare category freq
# BAR CHART


# Determine whether study hours are associated with GPA
# Variable(s): 2 quantitative
# Inspects relationship of two quantities
# Scatter plot
```

```python
import pandas as pd
import matplotlib.pyplot as plt

students = pd.DataFrame({
    'major': ['IS','CS','IS','SE','CS','IS','SE','CS'],
    'gender': ['F','M','F','M','F','M','F','M'],
    'age': [20,22,21,23,20,22,21,24],
    'GPA': [3.7,3.4,3.9,3.2,3.6,3.5,3.8,3.1],
    'study_hours': [8,4,10,3,7,6,9,2]
})


# Summarize the distribution of GPA - box plot
plt.boxplot(students['GPA'], vert=False)
plt.xlabel('GPA distribution')
plt.show()
print(" ")


# Show the distribution of ages across numerical intervals - histogram
weight= np.ones_like(students['age'])/students['age'].count()
plt.hist(students['age'], bins=5, weights=weight, edgecolor='black', facecolor='cyan')
plt.xlabel('Age')
plt.ylabel('Counts')
plt.show()
print(" ")

# Cumulative proportion of students at or below each GPA - ECDF
x= np.sort(students['GPA'])
y = np.linspace(0,1,len(students['GPA']))
plt.step(x,y)
plt.xlabel('GPA')
plt.ylabel('Cumulative Distribution')
plt.xlim(x.min(), x.max()) # provides explicit control over what portions of the dataset are displayed
# since step(x,y) alr passes values matplotlib can automatically determine the x-axis range
plt.show()
# Matplotlib can determine automatic limits by default, except that its format varies because it leaves margins around the data
# Explicit control overrides these automatic limits so the displayed axis correspond exactly to min+max
```

```python
# Given dataset

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

students = pd.DataFrame({
    'major': ['IS','CS','IS','SE','CS','IS','SE','CS','IS','SE'],
    'gender': ['Female','Male','Female','Male','Female',
               'Male','Female','Male','Female','Male'],
    'age': [20,22,21,23,20,22,21,24,23,21],
    'gpa': [3.7,3.4,3.9,3.2,3.6,3.5,3.8,3.1,3.7,3.3],
    'study_hours': [8,4,10,3,7,6,9,2,8,5]
})
```

```python
# The university wants to visualize how many students belong to each major (Category)

major_counts = students['major'].value_counts()
x= major_counts.index
plt.bar(x,major_counts, edgecolor='black', facecolor='cyan')
plt.xlabel('Major Categories')
plt.ylabel('Frequency')
plt.show()
```

```python
# Create a visualization that provides a compact statistical summary of the students' GPA distribution. Display it horizontally.
plt.boxplot(students['gpa'], vert=False)
plt.xlabel('GPA')
plt.show()
```

```python
# Visualize the distribution of students' ages across 5 numerical intervals, where the y-axis represents the proportion of total students, not raw counts
weights= np.ones_like(students['age'])/students['age'].count()
plt.hist(students['age'], weights=weights, bins=5, edgecolor='black', linewidth=0.5,facecolor='violet')
plt.xlabel('Age Range')
plt.ylabel('Proportion of Total')
plt.show()
```

```python
# Create a visualization showing the cumulative proportion of students at or below each GPA. Restrict the x-axis exactly to the observed GPA range

x = np.sort(students['gpa'])
y = np.linspace(0,1,len(students['gpa']))
plt.step(x,y)
plt.xlabel('GPA')
plt.ylabel('Cumulative Proportion')
plt.xlim(x.min(),x.max())
plt.show()
```

```python
# The university wants to compare the number of Male and Female students within each major.
# Use a Pandas approach that first constructs a table
# containing the combinations of the two categorical variables. - crosstab

count= pd.crosstab(students['major'], students['gender'])
count.plot(kind='bar', edgecolor='black')
plt.xlabel('Categories')
plt.ylabel('Frequency')
plt.show()
```

```python
# Seaborn approch
sns.countplot(x='major', hue='gender', data=students,
              hue_order=['Female', 'Male'],
              palette=['pink','skyblue'],
              saturation=2, edgecolor='black')

plt.legend(loc='upper right')
plt.xlabel('Categories')
plt.ylabel('Frequency')
plt.show()
```
